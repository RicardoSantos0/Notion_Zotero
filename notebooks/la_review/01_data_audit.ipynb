# 01 — Data Audit

Corpus composition and the PRISMA selection flow from the canonical bundles.

```
from collections import Counter
from notion_zotero.analysis import la_review_figure_data as fd
CANONICAL_DIR = fd._REPO_ROOT / "data" / "pulled" / "notion" / "learning_analytics_review"
from notion_zotero.analysis.paper_tables import _load_canonical_bundles, _bundle_task_label
bundles = _load_canonical_bundles(CANONICAL_DIR)
print("bundles:", len(bundles))
```

```
tasks = Counter(_bundle_task_label(b) for b in bundles)
years = Counter(fd._paper_year(b) for b in bundles if fd._paper_year(b))
print("by task:", dict(tasks))
print("year range:", min(years), "-", max(years))
```

```
prisma = fd.build_prisma_flow(CANONICAL_DIR)
print(prisma.to_string(index=False))
```
