# 03 — LA Review Figures (1-6)

Builds figure data from the canonical bundles and renders Figures 1-6 (PNG + Markdown caption). Runs clean from a fresh kernel; all logic lives in the tested package module `la_review_figure_data`.

```
from notion_zotero.analysis import la_review_figure_data as fd

CANONICAL_DIR = fd._REPO_ROOT / "data" / "pulled" / "notion" / "learning_analytics_review"
FIGURE_DATA_DIR = fd._LA_REVIEW_ROOT / "figure_data"
FIGURES_DIR = fd._LA_REVIEW_ROOT / "figures"
print("canonical dir exists:", CANONICAL_DIR.exists())
```

## 1. Build figure data (5 CSVs)

```
data = fd.build_all_figure_data(CANONICAL_DIR)
for name, df in data.items():
    print(f"{name}: {df.shape[0]} rows x {df.shape[1]} cols")
```

## 2. Render Figures 1-6 (PNG + captions)

```
pngs = fd.render_all_figures(FIGURE_DATA_DIR, FIGURES_DIR)
print("rendered:", [p.name for p in pngs])
```

## 3. Verify outputs

```
expected = [f"figure_{i}.png" for i in range(1, 7)] + [f"figure_{i}_caption.md" for i in range(1, 7)]
missing = [n for n in expected if not (FIGURES_DIR / n).exists()]
assert not missing, f"missing figure outputs: {missing}"
print("all figure outputs present")
```
